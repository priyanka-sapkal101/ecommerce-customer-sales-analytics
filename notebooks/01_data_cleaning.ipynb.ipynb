```python

# Select rows where Description is missing
missing_desc = clean_df.loc[clean_df["Description"].isna()]

print("Rows with missing descriptions:", len(missing_desc))

print(
    "Rows with a product code:",
    missing_desc["StockCode"].notna().sum()
)

display(missing_desc.head(10))
```

```python

# Build a lookup of known product descriptions
description_lookup = (
    clean_df.dropna(subset=["Description"])
    .groupby("StockCode")["Description"]
    .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else None)
)

# Check how many missing descriptions can be matched
missing_codes = missing_desc["StockCode"]

recoverable = missing_codes.isin(description_lookup.index)

print("Missing descriptions:", len(missing_desc))
print("Product codes with a known description:",
      recoverable.sum())
print("Product codes without a known description:",
      (~recoverable).sum())
```

```python

# Count missing descriptions before filling
before_missing = clean_df["Description"].isna().sum()

# Fill descriptions using the product-code lookup
clean_df["Description"] = clean_df["Description"].fillna(
    clean_df["StockCode"].map(description_lookup)
)

# Verify the result
after_missing = clean_df["Description"].isna().sum()

print("Missing descriptions before:", before_missing)
print("Descriptions filled:", before_missing - after_missing)
print("Missing descriptions remaining:", after_missing)
```

```python

import numpy as np

# Classify transactions without deleting any records
clean_df["TransactionType"] = np.select(
    [
        clean_df["Description"].str.contains(
            "adjust bad debt", case=False, na=False
        ),
        clean_df["Invoice"].astype(str).str.startswith("C"),
        clean_df["Price"] < 0,
        clean_df["Quantity"] < 0,
        clean_df["Price"] == 0
    ],
    [
        "Bad Debt Adjustment",
        "Cancellation Invoice",
        "Negative Price",
        "Negative Quantity",
        "Zero Price"
    ],
    default="Regular Transaction"
)

print(clean_df["TransactionType"].value_counts())
```

```python

# Create flags for transaction investigation
clean_df["IsCancellationInvoice"] = (
    clean_df["Invoice"].astype(str).str.startswith("C")
)

clean_df["IsNegativeQuantity"] = clean_df["Quantity"] < 0
clean_df["IsZeroPrice"] = clean_df["Price"] == 0
clean_df["IsNegativePrice"] = clean_df["Price"] < 0
clean_df["HasMissingCustomerID"] = clean_df["Customer ID"].isna()

# Summary of important data-quality flags
quality_summary = pd.DataFrame({
    "Issue": [
        "Cancellation invoice",
        "Negative quantity",
        "Zero price",
        "Negative price",
        "Missing Customer ID",
        "Negative quantity without C invoice"
    ],
    "Rows": [
        clean_df["IsCancellationInvoice"].sum(),
        clean_df["IsNegativeQuantity"].sum(),
        clean_df["IsZeroPrice"].sum(),
        clean_df["IsNegativePrice"].sum(),
        clean_df["HasMissingCustomerID"].sum(),
        (
            clean_df["IsNegativeQuantity"] &
            ~clean_df["IsCancellationInvoice"]
        ).sum()
    ]
})

quality_summary["Percentage"] = (
    quality_summary["Rows"] / len(clean_df) * 100
).round(2)

display(quality_summary)
```

```python

# Create a separate dataset for regular sales analysis
sales_df = clean_df[
    (clean_df["Quantity"] > 0) &
    (clean_df["Price"] > 0) &
    (~clean_df["Invoice"].astype(str).str.startswith("C"))
].copy()

# Calculate revenue
sales_df["Revenue"] = sales_df["Quantity"] * sales_df["Price"]

print("Original cleaned dataset:", len(clean_df))
print("Sales analysis dataset:", len(sales_df))
print("Total sales revenue:", round(sales_df["Revenue"].sum(), 2))
print("Missing customer IDs in sales data:", sales_df["Customer ID"].isna().sum())

display(sales_df.head())
```

```python

# Extract month from InvoiceDate
sales_df["Month"] = sales_df["InvoiceDate"].dt.to_period("M")

# Calculate monthly revenue
monthly_revenue = (
    sales_df.groupby("Month")
    .agg(
        Revenue=("Revenue", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Unique_Invoices=("Invoice", "nunique")
    )
    .reset_index()
)

monthly_revenue["Revenue"] = monthly_revenue["Revenue"].round(2)

display(monthly_revenue)
```

```python

import matplotlib.pyplot as plt

# Convert Month to strings for chart labels
monthly_revenue["Month_Label"] = monthly_revenue["Month"].astype(str)

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_revenue["Month_Label"],
    monthly_revenue["Revenue"],
    marker="o"
)

plt.title("Monthly Revenue Trend", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=90)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

total_revenue = sales_df["Revenue"].sum()
total_invoices = sales_df["Invoice"].nunique()

aov = total_revenue / total_invoices if total_invoices else 0

print("Total revenue:", round(total_revenue, 2))
print("Unique invoices:", total_invoices)
print("Average Order Value (AOV):", round(aov, 2))
```

```python

# Count distinct invoices per identified customer
customer_orders = (
    customer_df.groupby("Customer ID")["Invoice"]
    .nunique()
)

total_customers = len(customer_orders)

repeat_customers = (customer_orders >= 2).sum()

repeat_customer_rate = (
    repeat_customers / total_customers * 100
    if total_customers > 0 else 0
)

print("Total identified customers:", total_customers)
print("Repeat customers:", repeat_customers)
print("One-time customers:", (customer_orders == 1).sum())
print("Repeat customer rate (%):", round(repeat_customer_rate, 2))
```

```python

# Country-level sales analysis
country_analysis = (
    sales_df.groupby("Country")
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Unique_Invoices=("Invoice", "nunique")
    )
    .reset_index()
    .sort_values("Total_Revenue", ascending=False)
)

country_analysis["Total_Revenue"] = (
    country_analysis["Total_Revenue"].round(2)
)

print("Number of countries:", country_analysis["Country"].nunique())

display(country_analysis.head(10))
```

```python

# Extract calendar month number and name
sales_df["Month_Number"] = sales_df["InvoiceDate"].dt.month
sales_df["Month_Name"] = sales_df["InvoiceDate"].dt.month_name()

# Average revenue per calendar month across years
seasonality = (
    sales_df.groupby(["Month_Number", "Month_Name"])
    .agg(
        Average_Monthly_Revenue=("Revenue", "mean"),
        Total_Revenue=("Revenue", "sum"),
        Transaction_Rows=("Revenue", "count")
    )
    .reset_index()
    .sort_values("Month_Number")
)

# Calculate average total revenue by month across years
year_month_revenue = (
    sales_df.groupby(sales_df["InvoiceDate"].dt.to_period("M"))["Revenue"]
    .sum()
    .reset_index(name="Revenue")
)

year_month_revenue["Month_Number"] = (
    year_month_revenue["InvoiceDate"].dt.month
)

average_revenue_by_month = (
    year_month_revenue.groupby("Month_Number")["Revenue"]
    .mean()
    .reset_index(name="Average_Revenue_Across_Years")
)

display(average_revenue_by_month)
```

```python

import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

month_names = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

plt.figure(figsize=(12, 6))

plt.plot(
    average_revenue_by_month["Month_Number"],
    average_revenue_by_month["Average_Revenue_Across_Years"],
    marker="o",
    linewidth=2
)

plt.xticks(range(1, 13), month_names)
plt.title("Average Monthly Revenue Across Years", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Average Revenue")

plt.gca().yaxis.set_major_formatter(
    mticker.StrMethodFormatter("{x:,.0f}")
)

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```
